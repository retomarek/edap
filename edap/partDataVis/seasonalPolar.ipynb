# Polar Chart

```{admonition} Goal
:class: tip
Circular visualization of monthly energy demand, providing an intuitive view of the annual cycle.
```

```{admonition} Data Basis
:class: note
Monthly energy consumption data (`seasonal_monthly.csv`) bundled with the `pyedautils` package. 10 years of monthly values.
```

## Data Preparation

````python
import pandas as pd
from importlib import resources

data_path = resources.files("pyedautils") / "data" / "seasonal_monthly.csv"
df = pd.read_csv(data_path, sep=";")
df.head()
````

## Visualization

````python
from pyedautils.plots import plot_seasonal_polar

fig = plot_seasonal_polar(
    data=df,
    title="Seasonal Polar (10 Years)",
    ylab="Mean daily energy [kWh/d]",
)
fig.show()
````
