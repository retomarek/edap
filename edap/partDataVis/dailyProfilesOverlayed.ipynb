# Overlaid Profiles

```{admonition} Goal
:class: tip
Overlay all individual daily electricity consumption profiles. Hovering highlights the entire day-line in orange for comparison.
```

```{admonition} Data Basis
:class: note
Electricity data (`ele_meter.csv`) bundled with the `pyedautils` package: the mean electric power of a household in W per 15-minute interval (the 15-minute consumption in Wh × 4), for 2015.
```

## Data Preparation

````python
import pandas as pd
from importlib import resources

data_path = resources.files("pyedautils") / "data" / "ele_meter.csv"
df = pd.read_csv(data_path, sep=";")
df.head()
````

## Visualization

Note: The `method="overlayed"` option returns an HTML string instead of a Plotly Figure.

````python
from pyedautils.plots import plot_daily_profiles
from IPython.display import HTML

html = plot_daily_profiles(
    data=df,
    method="overlayed",
    title="Electricity - Overlaid Daily Profiles",
    ylab="Power [W]",
)
HTML(html)
````
