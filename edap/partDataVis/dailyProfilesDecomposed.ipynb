# Decomposed Profile

```{admonition} Goal
:class: tip
Display daily electricity consumption profiles with the trend component removed, isolating the intra-day pattern from long-term drift.
```

```{admonition} Data Basis
:class: note
Electricity data (`ele_meter.csv`) bundled with the `pyedautils` package: the mean electric power of a household in W per 15-minute interval (the 15-minute consumption in Wh × 4), for 2015.
```

## Data Preparation

````python
import pandas as pd
from importlib import resources

data_path = resources.files("pyedautils") / "data" / "ele_meter.csv"
df = pd.read_csv(data_path, sep=";")
df.head()
````

## Visualization

````python
from pyedautils.plots import plot_daily_profiles

fig = plot_daily_profiles(
    data=df,
    method="decomposed",
    loc_time_zone="Europe/Zurich",
    title="Electricity - Decomposed Weekly Pattern",
    ylab="Power deviation [W]",
)
fig.show()
````

::::{dropdown} Customization
The trend is a centered rolling mean over `k` rows. With 15-minute data the default `k=672` is one week, so the weekly pattern stays in the profile and slower changes (seasons, holidays) are removed. A window shorter than a day would also remove the daily pattern.

```python
fig = plot_daily_profiles(df, method="decomposed", loc_time_zone="Europe/Zurich", k=4 * 672)   # 4-week trend
```
