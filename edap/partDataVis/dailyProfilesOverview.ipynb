# Overview

```{admonition} Goal
:class: tip
Display a 4x7 grid of daily electricity consumption profiles, organized by season (rows) and weekday (columns), to reveal seasonal and weekly patterns.
```

```{admonition} Data Basis
:class: note
Electricity data (`ele_meter.csv`) bundled with the `pyedautils` package: the mean electric power of a household in W per 15-minute interval (the 15-minute consumption in Wh × 4), for 2015.
```

## Data Preparation

````python
import pandas as pd
from importlib import resources

data_path = resources.files("pyedautils") / "data" / "ele_meter.csv"
df = pd.read_csv(data_path, sep=";")
df.head()
````

## Visualization

````python
from pyedautils.plots import plot_daily_profiles_overview

fig = plot_daily_profiles_overview(
    data=df,
    title="Electricity Meter - Daily Profiles",
    ylab="Power [W]",
    confidence=95.0,
)
fig.show()
````

::::{dropdown} Customization
The `plot_daily_profiles_overview` function accepts the following optional parameters:

- `confidence`: Set the confidence interval band (e.g., `95.0` for 95%).
- `colors`: Customize the color scheme.

```python
fig = plot_daily_profiles_overview(df, title="Custom Example", ylab="Power [W]", confidence=90.0)
```
